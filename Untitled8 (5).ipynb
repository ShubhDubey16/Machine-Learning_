```python
X= df.drop('Machine failure', axis=1)
```

```python
y= df["Machine failure"]
```

```python
print("Input shape :",X.shape)
print("Output shape :",y.shape)
```

```python
categorical_features=["Type"]

numerical_features=[
    "Air temperature [K]",
    "Process temperature [K]",
    "Rotational speed [rpm]",
    "Torque [Nm]",
    "Tool wear [min]"
]
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.20,random_state=42,stratify=y)

print("Training samples :",X_train.shape)
print("Testing samples :",X_test.shape)
```

```python
preprocessor=ColumnTransformer(
    transformers=[
        ("num",StandardScaler(),numerical_features),
        ("cat",OneHotEncoder(handle_unknown="ignore"),categorical_features)
    ]
)
```

**Kernel Linear**

```python
linear_svm = Pipeline(
    steps=[("preprocessor",preprocessor),("classifier",SVC(kernel="linear"))]

)

linear_svm.fit(X_train,y_train)
```

```python
y_pred_linear = linear_svm.predict(X_test)
```

```python
accuracy_linear = accuracy_score(y_test,y_pred_linear)
precision_linear = precision_score(y_test,y_pred_linear,zero_division=0)
recall_linear = recall_score(y_test,y_pred_linear,zero_division=0)
f1_linear = f1_score(y_test,y_pred_linear,zero_division=0)

print("Accuracy :",accuracy_linear)
print("Precision :",precision_linear)
print("Recall :",recall_linear)
print("F1 Score :",f1_linear)
```

```python
print(
    classification_report(
        y_test,
        y_pred_linear,
        target_names=["No failure"," failure"],zero_division=0
    )
)
```

```python
cm_linear=confusion_matrix(y_test,y_pred_linear)
plt.figure(figsize=(7,4))
sns.heatmap(cm_linear,annot=True,fmt="d",cmap="summer",
            xticklabels=["No failure","Failure"])
plt.xlabel("predicted")
plt.ylabel("actual")
plt.title("Confusion Matrix - Linear SVM")
plt.show()
```

**Kernel  - RBF**

```python
rbf_svm = Pipeline(
    steps=[("preprocessor",preprocessor),("classifier",SVC(kernel="rbf"))]

)

rbf_svm.fit(X_train,y_train)
```

```python
y_pred_rbf = rbf_svm.predict(X_test)
```

```python
accuracy_rbf = accuracy_score(y_test,y_pred_rbf)
precision_rbf = precision_score(y_test,y_pred_rbf,zero_division=0)
recall_rbf = recall_score(y_test,y_pred_rbf,zero_division=0)
f1_rbf = f1_score(y_test,y_pred_rbf,zero_division=0)

print("Accuracy :",accuracy_rbf)
print("Precision :",precision_rbf)
print("Recall :",recall_rbf)
print("F1 Score :",f1_rbf)
```

```python
print(
    classification_report(
        y_test,
        y_pred_rbf,
        target_names=["No failure"," failure"],zero_division=0
    )
)
```

```python
cm_rbf=confusion_matrix(y_test,y_pred_rbf)
plt.figure(figsize=(7,4))
sns.heatmap(cm_rbf,annot=True,fmt="d",cmap="summer",
            xticklabels=["No failure","Failure"])
plt.xlabel("predicted")
plt.ylabel("actual")
plt.title("Confusion Matrix - RBF SVM")
plt.show()
```

**Kernel - Polynomial**

```python
poly_svm = Pipeline(
    steps=[("preprocessor",preprocessor),("classifier",SVC(kernel="poly"))]

)

poly_svm.fit(X_train,y_train)
```

```python
y_pred_poly = poly_svm.predict(X_test)
```

```python
accuracy_poly = accuracy_score(y_test,y_pred_poly)
precision_poly = precision_score(y_test,y_pred_poly,zero_division=0)
recall_poly = recall_score(y_test,y_pred_poly,zero_division=0)
f1_poly = f1_score(y_test,y_pred_poly,zero_division=0)

print("Accuracy :",accuracy_poly)
print("Precision :",precision_poly)
print("Recall :",recall_poly)
print("F1 Score :",f1_poly)
```

```python
print(
    classification_report(
        y_test,
        y_pred_poly,
        target_names=["No failure"," failure"],zero_division=0
    )
)
```

```python
cm_poly=confusion_matrix(y_test,y_pred_poly)
plt.figure(figsize=(7,4))
sns.heatmap(cm_poly,annot=True,fmt="d",cmap="summer",
            xticklabels=["No failure","Failure"])
plt.xlabel("predicted")
plt.ylabel("actual")
plt.title("Confusion Matrix - Poly SVM")
plt.show()
```

```python
comparison=pd.DataFrame({
    "Kernel":["Linear","RBF","Poly"],
    "Accuracy":[accuracy_linear,accuracy_rbf,accuracy_poly],
    "Precision":[precision_linear,precision_rbf,precision_poly],
    "Recall":[recall_linear,recall_rbf,recall_poly],
    "F1 Score":[f1_linear,f1_rbf,f1_poly]
})
comparison
```

```python
comparison_plot=comparison.set_index("Kernel")
comparison_plot.plot(kind="bar",figsize=(10,5))
plt.title("Comparison of Kernels")
plt.xlabel("SVM Kernel")
plt.ylabel("Score")
plt.ylim(0,1)
plt.xticks(rotation=0)
plt.legend(bbox_to_anchor=(1.05,1),loc="upper left")
plt.tight_layout()
plt.show()
```

```python
from sklearn.pipeline import Pipeline
pipeline = Pipeline(
    steps=[("preprocessor",preprocessor),("classifier",SVC())]
)
```

```python
param_grid=[{"classifier_kernel" : ["linear"],"classifier_C":[0.1,1,10,100]},
            {"classifier_kernel":["rbf"],"classifier_C":[0.1,1,10,100],"classifier_gamma":["scale",0.01,0.1,1]}]
```

```python
grid_search=GridSearchCV(estimator=pipeline,param_grid=param_grid,cv=5,scoring="f1",n_jobs=-1,verbose=1)
```

```python

correct_param_grid = [
    {"classifier__kernel": ["linear"], "classifier__C": [0.1, 1, 10, 100]},
    {
        "classifier__kernel": ["rbf"],
        "classifier__C": [0.1, 1, 10, 100],
        "classifier__gamma": ["scale", 0.01, 0.1, 1]
    }
]


grid_search = GridSearchCV(
    estimator=pipeline,
    param_grid=correct_param_grid,
    cv=5,
    scoring="f1",
    n_jobs=-1,
    verbose=1
)


print("Fitting GridSearchCV on training data...")
grid_search.fit(X_train, y_train)

print("\nBest Parameters : ")
print(grid_search.best_params_)
```

```python
print("Best Cross Validation F1 Score")
```

```python
best_model=grid_search.best_estimator_
y_pred_tuned=best_model.predict(X_test)
```

```python
accuracy_tuned=accuracy_score(y_test,y_pred_tuned)
precision_tuned=precision_score(y_test,y_pred_tuned,zero_division=0)
recall_tuned=recall_score(y_test,y_pred_tuned,zero_division=0)
f1_tuned=f1_score(y_test,y_pred_tuned,zero_division=0)

print("Tuned SVM Results")
print("Accuracy : ",accuracy_tuned)
print("Precision : ",precision_tuned)
print("Recall : ",recall_tuned)
print("F1 Score : ",f1_tuned)
```
